```python
X_train,X_test,y_train,y_test = train_test_split(df.iloc[:,1:],df.iloc[:,0],test_size=0.2,random_state=42)
```

### Numpy Array to Pytorch Tensors

```python
X_train_tensor = torch.from_numpy(X_train)
X_test_tensor = torch.from_numpy(X_test)
y_train_tensor = torch.from_numpy(y_train)
y_test_tensor = torch.from_numpy(y_test)
```

### We have to define a class

```python
class MySimpleNN():
  def __init__(self,X):
    self.weights = torch.rand(X.shape[1],1,dtype=torch.float64, requires_grad=True)
    self.bias = torch.rand(1,dtype=torch.float64, requires_grad=True)

  def forward(self,X):
    z = torch.matmul(X,self.weights) + self.bias
    y_pred = torch.sigmoid(z)
    return y_pred

  def loss_function(self,y_pred,y):
    epsilon = 1e-7
    y_pred = torch.clamp(y_pred,epsilon,1-epsilon)
    loss = -(y_train_tensor * torch.log(y_pred) + (1-y_train_tensor) * torch.log(1-y_pred)).mean()
    return loss
```

```python
learning_rate = 0.001
epochs = 25
```

```python
model = MySimpleNN(X_test_tensor)
model.weights
```

```python
model.bias
```

### We have to train our model

```python
# 3 Steps are there
# Define a loop
model = MySimpleNN(X_train_tensor)
for epoch in range(epochs):

  # Forward Pass
  y_pred = model.forward(X_train_tensor)

  # Loss Calculate
  loss = model.loss_function(y_pred,y_train_tensor)
  print(f"Epoch: {epoch+1}, Loss: {loss.item()}")

  # Backward pass
  loss.backward()

  # Parameters Update
  with torch.no_grad():
    model.weights -= learning_rate * model.weights.grad
    model.bias -= learning_rate * model.bias.grad

  # Backward pass
  model.weights.grad.zero_()
  model.bias.grad.zero_()

  #Print loss in each epoch
  print(f"Epoch: {epoch+1}, Loss: {loss.item()}")
```

```python
import torch.nn as nn

class MySimpleNN(nn.Module):
  def __init__(self,num_features):
    super().__init__()
    self.linear = nn.Linear(num_features,1)
    self.sigmoid = nn.Sigmoid()

  def forward(self,features):
    out = self.linear(features)
    out = self.sigmoid(out)
    return out

learning_rate = 0.08
epochs = 25

loss_function = nn.BCELoss()

# Convert inputs to Float (float32) to match PyTorch layer defaults
X_train_tensor = X_train_tensor.float()
y_train_tensor = y_train_tensor.float()

# 3 Steps are there
# Define a loop
model = MySimpleNN(X_train_tensor.shape[1])

#Define an optimiser
optimizer = torch.optim.SGD(model.parameters(),lr=learning_rate)

for epoch in range(epochs):

  # Forward Pass
  y_pred = model.forward(X_train_tensor)

  # Loss Calculate
  loss = loss_function(y_pred,y_train_tensor.view(-1,1))
  print(f"Epoch: {epoch+1}, Loss: {loss.item()}")

  # Backward pass
  optimizer.zero_grad()

  # Backward pass
  loss.backward()

  # Parameters Update
  optimizer.step()

  #Print loss in each epoch
  print(f"Epoch: {epoch+1}, Loss: {loss.item()}")
```
